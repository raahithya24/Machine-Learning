```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'Price': [80, 120, 150, 200, 250, 300, 180, 220, 100, 270],
    'Area': [900, 1200, 1500, 1800, 2200, 2500, 1600, 2000, 1000, 2300],
    'Bedrooms': [2, 3, 3, 4, 4, 5, 3, 4, 2, 5],
    'Age': [5, 8, 10, 15, 20, 25, 12, 18, 6, 22],
    'Distance': [5, 8, 10, 12, 15, 18, 9, 14, 6, 16]
}

df1 = pd.DataFrame(data)

print(df1)
```

```python
sc = StandardScaler()

sc.fit(df1)

scaled_data = sc.transform(df1)

principal = PCA(n_components=3)

principal.fit(scaled_data)

x = principal.transform(scaled_data)

print(x.shape)
```

```python
print("Principal Components:")
print(principal.components_)
```

```python
print("Principal Components:")
print(principal.components_)
```

```python
plt.figure(figsize=(10,10))

plt.scatter(
    x[:,0],
    x[:,1]
)

plt.title('PCA 2D - Real Estate Properties')
plt.xlabel('PC1')
plt.ylabel('PC2')

plt.show()
```

```python
features = df1.columns

for i, component in enumerate(principal.components_):
    
    print("PC", i+1)
    
    for feature, loading in zip(features, component):
        print(feature, ":", round(loading, 3))
    
    print()
```
